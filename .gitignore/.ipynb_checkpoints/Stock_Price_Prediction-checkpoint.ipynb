# Stock Price Prediction Project

**Objective:** Analyze and predict stock prices using historical data and machine learning techniques.

**Libraries Used:** pandas, yfinance, numpy, matplotlib, seaborn, scikit-learn

**Stocks Analyzed:** ETERNAL.NS, GODFRYPHLP.NS, ONGC.NS, RELIANCE.NS, WIPRO.NS

## 1. Import Libraries and Setup

```python
# Data manipulation and analysis
import yfinance as yf
import numpy as np
import pandas as pd

# Visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Machine Learning
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Settings
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")
pd.set_option('display.max_columns', None)

import warnings
warnings.filterwarnings('ignore')
```

## 2. Data Collection

```python
# Define stock symbols and date range
stocks = ["ETERNAL.NS", "GODFRYPHLP.NS", "ONGC.NS", "RELIANCE.NS", "WIPRO.NS"]
start_date = "2020-01-01"
end_date = "2024-12-31"

# Download data
print("Downloading stock data...")
data = yf.download(stocks, start=start_date, end=end_date)

print(f"\nData shape: {data.shape}")
print(f"Date range: {data.index[0]} to {data.index[-1]}")
data.head()
```

```python
# Check for missing values
print("Missing values per stock:")
print(data['Close'].isnull().sum())

# Fill missing values using forward fill
data = data.fillna(method='ffill')
```

```python
# Basic statistics
print("\n=== DESCRIPTIVE STATISTICS ===")
data['Close'].describe()
```

## 3. Exploratory Data Analysis (EDA)

### 3.1 Price Trends Over Time

```python
# Extract individual stock closing prices
eternal = data["Close"]["ETERNAL.NS"]
godfry = data["Close"]["GODFRYPHLP.NS"]
ongc = data["Close"]["ONGC.NS"]
reliance = data["Close"]["RELIANCE.NS"]
wipro = data["Close"]["WIPRO.NS"]

# Plot all stocks
plt.figure(figsize=(16, 8))
plt.plot(eternal, label='Eternal', linewidth=2)
plt.plot(godfry, label='Godfrey Phillips', linewidth=2)
plt.plot(ongc, label='ONGC', linewidth=2)
plt.plot(reliance, label='Reliance', linewidth=2)
plt.plot(wipro, label='Wipro', linewidth=2)

plt.title('Stock Price Comparison (2020-2024)', fontsize=16, fontweight='bold')
plt.xlabel('Date', fontsize=12)
plt.ylabel('Closing Price (₹)', fontsize=12)
plt.legend(loc='best', fontsize=11)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### 3.3 Volume Analysis

```python
# Volume comparison
plt.figure(figsize=(16, 6))
data['Volume'].plot(kind='line', alpha=0.7)
plt.title('Trading Volume Comparison', fontsize=16, fontweight='bold')
plt.xlabel('Date', fontsize=12)
plt.ylabel('Volume', fontsize=12)
plt.legend(stocks, loc='best')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### 3.4 Daily Returns Analysis

```python
# Calculate daily returns
ete_return = eternal.pct_change()
gfp_return = godfry.pct_change()
ongc_return = ongc.pct_change()
reliance_return = reliance.pct_change()
wipro_return = wipro.pct_change()

# Combine into DataFrame
daily_returns = pd.DataFrame({
    'ETERNAL': ete_return,
    'GODFRYPHLP': gfp_return,
    'ONGC': ongc_return,
    'RELIANCE': reliance_return,
    'WIPRO': wipro_return
})

print("Daily Returns Statistics:")
daily_returns.describe()
```

```python
# Distribution of daily returns
fig, axes = plt.subplots(2, 3, figsize=(18, 10))
fig.suptitle('Daily Returns Distribution', fontsize=16, fontweight='bold')

stocks_list = daily_returns.columns
colors = ['#E63946', '#F4A261', '#2A9D8F', '#264653', '#E76F51']

for idx, (stock, ax) in enumerate(zip(stocks_list, axes.flatten())):
    daily_returns[stock].hist(bins=50, alpha=0.7, color=colors[idx], ax=ax)
    ax.axvline(daily_returns[stock].mean(), color='red', linestyle='--', linewidth=2, label='Mean')
    ax.set_title(stock, fontsize=12, fontweight='bold')
    ax.set_xlabel('Daily Return')
    ax.set_ylabel('Frequency')
    ax.legend()
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

### 3.5 Correlation Analysis

```python
# Correlation heatmap
plt.figure(figsize=(10, 8))
correlation_matrix = daily_returns.corr()
sns.heatmap(correlation_matrix, annot=True, cmap="YlGnBu", fmt=".2f", 
            linewidths=0.5, cbar_kws={"shrink": 0.8})
plt.title("Correlation Between Stock Daily Returns", fontsize=16, fontweight='bold', pad=20)
plt.tight_layout()
plt.show()

print("\nKey Insights:")
print(f"Highest correlation: {correlation_matrix.unstack().sort_values(ascending=False).drop_duplicates()[1]:.3f}")
print(f"Lowest correlation: {correlation_matrix.unstack().sort_values(ascending=True)[0]:.3f}")
```

### 3.6 Risk vs Return Analysis

```python
# Calculate risk (std) and return (mean)
risks = daily_returns.std()
returns = daily_returns.mean()

plt.figure(figsize=(12, 7))
plt.scatter(risks, returns, c="#DA4167", s=200, alpha=0.7, edgecolors='black', linewidth=2)

for stock in daily_returns.columns:
    plt.annotate(stock, (risks[stock], returns[stock]), 
                fontsize=11, fontweight='bold', 
                xytext=(10, 10), textcoords='offset points')

plt.title('Risk vs Return Analysis', fontsize=16, fontweight='bold')
plt.xlabel('Risk (Standard Deviation)', fontsize=12)
plt.ylabel('Expected Return (Mean)', fontsize=12)
plt.axhline(0, color='black', linewidth=0.5, linestyle='--')
plt.axvline(risks.mean(), color='green', linewidth=1, linestyle='--', alpha=0.5, label='Avg Risk')
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

### 3.7 Sharpe Ratio Ranking

```python
# Sharpe Ratio calculation
risk_free_annual = 0.0693  # 6.93% - RBI risk-free rate
risk_free_daily = risk_free_annual / 252  # 252 trading days

sharpe_ratio = (returns - risk_free_daily) / risks

print("=== SHARPE RATIO RANKING (Best to Worst) ===")
print(sharpe_ratio.sort_values(ascending=False))
print("\n" + "="*50)

print("\nAverage Daily Returns:")
print(returns)

print("\nAnnualized Returns (%):")
print(returns * 252 * 100)

print("\nRisk (Volatility - Daily Std Dev):")
print(risks)

print("\nRisk Free Annual Rate: 6.93%")
```

```python
# Visualize Sharpe Ratios
plt.figure(figsize=(12, 6))
sharpe_sorted = sharpe_ratio.sort_values(ascending=True)
colors_bar = ['#06D6A0' if x > 0 else '#EF476F' for x in sharpe_sorted]

sharpe_sorted.plot(kind='barh', color=colors_bar, edgecolor='black', linewidth=1.5)
plt.title('Sharpe Ratio Comparison (Risk-Adjusted Returns)', fontsize=16, fontweight='bold')
plt.xlabel('Sharpe Ratio', fontsize=12)
plt.ylabel('Stock', fontsize=12)
plt.axvline(0, color='black', linewidth=1)
plt.grid(True, alpha=0.3, axis='x')
plt.tight_layout()
plt.show()

print(f"\n🏆 Best Risk-Adjusted Stock: {sharpe_ratio.idxmax()} (Sharpe: {sharpe_ratio.max():.3f})")
```

## 4. Technical Indicators

We'll focus on **RELIANCE.NS** for prediction modeling as it's a well-known, liquid stock.

```python
# Select one stock for prediction (RELIANCE)
stock_symbol = "RELIANCE.NS"
df = data['Close'][stock_symbol].to_frame()
df.columns = ['Close']
df['Open'] = data['Open'][stock_symbol]
df['High'] = data['High'][stock_symbol]
df['Low'] = data['Low'][stock_symbol]
df['Volume'] = data['Volume'][stock_symbol]

print(f"Working with: {stock_symbol}")
print(f"Data points: {len(df)}")
df.head()
```

### 4.1 Moving Averages

```python
# Simple Moving Averages
df['SMA_10'] = df['Close'].rolling(window=10).mean()
df['SMA_20'] = df['Close'].rolling(window=20).mean()
df['SMA_50'] = df['Close'].rolling(window=50).mean()

# Exponential Moving Average
df['EMA_12'] = df['Close'].ewm(span=12, adjust=False).mean()
df['EMA_26'] = df['Close'].ewm(span=26, adjust=False).mean()

print("Moving Averages calculated.")
df[['Close', 'SMA_10', 'SMA_20', 'SMA_50', 'EMA_12']].tail(10)
```

### 4.2 Bollinger Bands

```python
# Bollinger Bands
df['BB_Middle'] = df['Close'].rolling(window=20).mean()
df['BB_Std'] = df['Close'].rolling(window=20).std()
df['BB_Upper'] = df['BB_Middle'] + (2 * df['BB_Std'])
df['BB_Lower'] = df['BB_Middle'] - (2 * df['BB_Std'])

print("Bollinger Bands calculated.")
```

### 4.3 RSI (Relative Strength Index)

```python
# RSI Calculation
def calculate_rsi(data, window=14):
    delta = data.diff()
    gain = (delta.where(delta > 0, 0)).rolling(window=window).mean()
    loss = (-delta.where(delta < 0, 0)).rolling(window=window).mean()
    rs = gain / loss
    rsi = 100 - (100 / (1 + rs))
    return rsi

df['RSI'] = calculate_rsi(df['Close'])
print("RSI calculated.")
print(f"Current RSI: {df['RSI'].iloc[-1]:.2f}")
print("RSI > 70: Overbought | RSI < 30: Oversold")
```

### 4.4 MACD (Moving Average Convergence Divergence)

```python
# MACD
df['MACD'] = df['EMA_12'] - df['EMA_26']
df['Signal_Line'] = df['MACD'].ewm(span=9, adjust=False).mean()
df['MACD_Histogram'] = df['MACD'] - df['Signal_Line']

print("MACD calculated.")
df[['MACD', 'Signal_Line', 'MACD_Histogram']].tail()
```

### 4.5 Visualize Technical Indicators

```python
# Plot Technical Indicators
fig, axes = plt.subplots(4, 1, figsize=(16, 14), sharex=True)

# Price with Moving Averages and Bollinger Bands
axes[0].plot(df.index, df['Close'], label='Close Price', color='black', linewidth=2)
axes[0].plot(df.index, df['SMA_20'], label='SMA 20', linestyle='--', alpha=0.7)
axes[0].plot(df.index, df['SMA_50'], label='SMA 50', linestyle='--', alpha=0.7)
axes[0].fill_between(df.index, df['BB_Upper'], df['BB_Lower'], alpha=0.2, label='Bollinger Bands')
axes[0].set_ylabel('Price (₹)', fontsize=11)
axes[0].set_title(f'{stock_symbol} - Technical Analysis', fontsize=14, fontweight='bold')
axes[0].legend(loc='best')
axes[0].grid(True, alpha=0.3)

# Volume
axes[1].bar(df.index, df['Volume'], alpha=0.5, color='steelblue')
axes[1].set_ylabel('Volume', fontsize=11)
axes[1].set_title('Trading Volume', fontsize=12, fontweight='bold')
axes[1].grid(True, alpha=0.3)

# RSI
axes[2].plot(df.index, df['RSI'], color='purple', linewidth=2)
axes[2].axhline(70, color='red', linestyle='--', alpha=0.7, label='Overbought (70)')
axes[2].axhline(30, color='green', linestyle='--', alpha=0.7, label='Oversold (30)')
axes[2].fill_between(df.index, 30, 70, alpha=0.1, color='gray')
axes[2].set_ylabel('RSI', fontsize=11)
axes[2].set_title('Relative Strength Index (RSI)', fontsize=12, fontweight='bold')
axes[2].legend(loc='best')
axes[2].grid(True, alpha=0.3)

# MACD
axes[3].plot(df.index, df['MACD'], label='MACD', color='blue', linewidth=2)
axes[3].plot(df.index, df['Signal_Line'], label='Signal Line', color='red', linewidth=2)
axes[3].bar(df.index, df['MACD_Histogram'], label='Histogram', alpha=0.3, color='gray')
axes[3].axhline(0, color='black', linewidth=0.5)
axes[3].set_ylabel('MACD', fontsize=11)
axes[3].set_xlabel('Date', fontsize=11)
axes[3].set_title('MACD (Moving Average Convergence Divergence)', fontsize=12, fontweight='bold')
axes[3].legend(loc='best')
axes[3].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## 5. Feature Engineering for Prediction

```python
# Create lag features (previous days' prices)
df['Lag_1'] = df['Close'].shift(1)
df['Lag_2'] = df['Close'].shift(2)
df['Lag_3'] = df['Close'].shift(3)
df['Lag_5'] = df['Close'].shift(5)

# Price change features
df['Daily_Return'] = df['Close'].pct_change()
df['Price_Change'] = df['Close'].diff()

# Rolling statistics
df['Rolling_Mean_7'] = df['Close'].rolling(window=7).mean()
df['Rolling_Std_7'] = df['Close'].rolling(window=7).std()
df['Rolling_Min_7'] = df['Close'].rolling(window=7).min()
df['Rolling_Max_7'] = df['Close'].rolling(window=7).max()

# High-Low range
df['HL_Range'] = df['High'] - df['Low']
df['OC_Change'] = df['Close'] - df['Open']

# Target variable: Next day's closing price
df['Target'] = df['Close'].shift(-1)

print("Feature engineering completed.")
print(f"Total features: {df.shape[1]}")
df.head()
```

```python
# Remove NaN values created by lagging/rolling
df_clean = df.dropna()
print(f"Clean dataset size: {len(df_clean)} rows")
print(f"Removed {len(df) - len(df_clean)} rows with missing values")
```

## 6. Stock Price Prediction Models

### 6.1 Prepare Data for Machine Learning

```python
# Select features for prediction
feature_columns = ['Open', 'High', 'Low', 'Volume', 
                   'SMA_10', 'SMA_20', 'SMA_50', 
                   'EMA_12', 'EMA_26', 'RSI', 'MACD',
                   'Lag_1', 'Lag_2', 'Lag_3', 'Lag_5',
                   'Rolling_Mean_7', 'Rolling_Std_7',
                   'HL_Range', 'OC_Change']

X = df_clean[feature_columns]
y = df_clean['Target']

print(f"Features (X): {X.shape}")
print(f"Target (y): {y.shape}")
```

```python
# Train-Test Split (80-20)
# Use time-series split (not random) to prevent data leakage
split_index = int(len(X) * 0.8)

X_train = X[:split_index]
X_test = X[split_index:]
y_train = y[:split_index]
y_test = y[split_index:]

print(f"Training set: {X_train.shape[0]} samples")
print(f"Test set: {X_test.shape[0]} samples")
print(f"Train period: {X_train.index[0]} to {X_train.index[-1]}")
print(f"Test period: {X_test.index[0]} to {X_test.index[-1]}")
```

### 6.2 Model 1: Linear Regression

```python
# Train Linear Regression model
lr_model = LinearRegression()
lr_model.fit(X_train, y_train)

# Predictions
y_pred_lr = lr_model.predict(X_test)

# Evaluation
mae_lr = mean_absolute_error(y_test, y_pred_lr)
rmse_lr = np.sqrt(mean_squared_error(y_test, y_pred_lr))
r2_lr = r2_score(y_test, y_pred_lr)

print("=== LINEAR REGRESSION RESULTS ===")
print(f"Mean Absolute Error (MAE): ₹{mae_lr:.2f}")
print(f"Root Mean Squared Error (RMSE): ₹{rmse_lr:.2f}")
print(f"R² Score: {r2_lr:.4f}")
print(f"Accuracy: {r2_lr * 100:.2f}%")
```

### 6.3 Model 2: Random Forest Regressor

```python
# Train Random Forest model
rf_model = RandomForestRegressor(n_estimators=100, max_depth=10, random_state=42, n_jobs=-1)
rf_model.fit(X_train, y_train)

# Predictions
y_pred_rf = rf_model.predict(X_test)

# Evaluation
mae_rf = mean_absolute_error(y_test, y_pred_rf)
rmse_rf = np.sqrt(mean_squared_error(y_test, y_pred_rf))
r2_rf = r2_score(y_test, y_pred_rf)

print("=== RANDOM FOREST RESULTS ===")
print(f"Mean Absolute Error (MAE): ₹{mae_rf:.2f}")
print(f"Root Mean Squared Error (RMSE): ₹{rmse_rf:.2f}")
print(f"R² Score: {r2_rf:.4f}")
print(f"Accuracy: {r2_rf * 100:.2f}%")
```

### 6.4 Feature Importance (Random Forest)

```python
# Feature importance
feature_importance = pd.DataFrame({
    'Feature': feature_columns,
    'Importance': rf_model.feature_importances_
}).sort_values('Importance', ascending=False)

plt.figure(figsize=(12, 8))
sns.barplot(data=feature_importance.head(15), x='Importance', y='Feature', palette='viridis')
plt.title('Top 15 Most Important Features for Prediction', fontsize=14, fontweight='bold')
plt.xlabel('Importance Score', fontsize=12)
plt.ylabel('Feature', fontsize=12)
plt.tight_layout()
plt.show()

print("\nTop 5 Most Important Features:")
print(feature_importance.head())
```

## 7. Model Evaluation & Comparison

```python
# Comparison table
comparison_df = pd.DataFrame({
    'Model': ['Linear Regression', 'Random Forest'],
    'MAE (₹)': [mae_lr, mae_rf],
    'RMSE (₹)': [rmse_lr, rmse_rf],
    'R² Score': [r2_lr, r2_rf],
    'Accuracy (%)': [r2_lr * 100, r2_rf * 100]
})

print("\n=== MODEL COMPARISON ===")
print(comparison_df.to_string(index=False))
print("\n" + "="*60)

best_model = comparison_df.loc[comparison_df['R² Score'].idxmax(), 'Model']
print(f"\n🏆 Best Model: {best_model}")
```

### 7.1 Visualization: Actual vs Predicted Prices

```python
# Plot predictions
plt.figure(figsize=(18, 8))

plt.plot(y_test.index, y_test.values, label='Actual Price', 
         color='black', linewidth=2.5, alpha=0.8)
plt.plot(y_test.index, y_pred_lr, label='Linear Regression Prediction', 
         linestyle='--', linewidth=2, alpha=0.7)
plt.plot(y_test.index, y_pred_rf, label='Random Forest Prediction', 
         linestyle='--', linewidth=2, alpha=0.7)

plt.title(f'{stock_symbol} - Actual vs Predicted Prices (Test Set)', 
          fontsize=16, fontweight='bold')
plt.xlabel('Date', fontsize=12)
plt.ylabel('Price (₹)', fontsize=12)
plt.legend(loc='best', fontsize=11)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### 7.2 Prediction Error Analysis

```python
# Error analysis
error_lr = y_test - y_pred_lr
error_rf = y_test - y_pred_rf

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Linear Regression errors
axes[0].hist(error_lr, bins=30, color='#E63946', alpha=0.7, edgecolor='black')
axes[0].axvline(0, color='black', linestyle='--', linewidth=2)
axes[0].set_title('Linear Regression - Prediction Errors', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Error (₹)', fontsize=11)
axes[0].set_ylabel('Frequency', fontsize=11)
axes[0].grid(True, alpha=0.3)

# Random Forest errors
axes[1].hist(error_rf, bins=30, color='#06D6A0', alpha=0.7, edgecolor='black')
axes[1].axvline(0, color='black', linestyle='--', linewidth=2)
axes[1].set_title('Random Forest - Prediction Errors', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Error (₹)', fontsize=11)
axes[1].set_ylabel('Frequency', fontsize=11)
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print(f"Linear Regression - Mean Error: ₹{error_lr.mean():.2f}")
print(f"Random Forest - Mean Error: ₹{error_rf.mean():.2f}")
```

## 8. Future Price Prediction

```python
# Predict next day's price using the latest data
latest_features = X.iloc[-1:]

next_day_pred_lr = lr_model.predict(latest_features)[0]
next_day_pred_rf = rf_model.predict(latest_features)[0]
current_price = df_clean['Close'].iloc[-1]

print("=== NEXT DAY PRICE PREDICTION ===")
print(f"Current Price (Last Known): ₹{current_price:.2f}")
print(f"\nLinear Regression Prediction: ₹{next_day_pred_lr:.2f}")
print(f"Change: ₹{next_day_pred_lr - current_price:.2f} ({((next_day_pred_lr/current_price - 1) * 100):.2f}%)")
print(f"\nRandom Forest Prediction: ₹{next_day_pred_rf:.2f}")
print(f"Change: ₹{next_day_pred_rf - current_price:.2f} ({((next_day_pred_rf/current_price - 1) * 100):.2f}%)")

avg_prediction = (next_day_pred_lr + next_day_pred_rf) / 2
print(f"\nAverage Prediction: ₹{avg_prediction:.2f}")
print(f"Change: ₹{avg_prediction - current_price:.2f} ({((avg_prediction/current_price - 1) * 100):.2f}%)")
```
