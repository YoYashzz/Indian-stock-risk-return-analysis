# Data loading

```python
import yfinance as yf
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
stocks=["ETERNAL.NS","GODFRYPHLP.NS","ONGC.NS","RELIANCE.NS","WIPRO.NS"]
data = yf.download(stocks,start = "2024-04-01", end="2026-03-30")
print(data.head())
```

```python
data["Close"]
```

```python
data.describe()
```

# Multi stock comparison


```python
wipro = data["Close"]["WIPRO.NS"]

plt.plot(wipro, color='#4E4187', label='Wipro Close Price')
plt.axhline(wipro.mean(),color = "red",linestyle = '--', label=('Avg Price'))
plt.tight_layout()
plt.title("Mean Value Of Wipro Closing Price")
plt.xlabel("Year")
plt.ylabel("Price (in INR)")
plt.legend()
plt.show()
```

```python
eternal = data["Close"]["ETERNAL.NS"]
godfry = data["Close"]["GODFRYPHLP.NS"]
ongc = data["Close"]["ONGC.NS"]
reliance= data["Close"]["RELIANCE.NS"]
wipro = data["Close"]["WIPRO.NS"]

plt.figure(figsize=(14, 6))
plt.plot(eternal, color = '#B80C09', label='ETERNAL')
plt.plot(godfry, color = '#6B2B06', label='GODFRYPHLP')
plt.plot(ongc, color = '#4A7856', label='ONGC')
plt.plot(reliance, color = '#083D77', label='RELIANCE')
plt.plot(wipro, color = '#141301', label='WIPRO')
plt.title("Price Comparison of Muliple Stocks")
plt.xlabel("Date")
plt.ylabel("Price")
plt.tight_layout()
plt.legend()
plt.show()
```

```python
wipro = data["Close"]["WIPRO.NS"]
plt.figure(figsize=(14,6))
plt.plot(wipro, color = '#DB504A', label='WIPRO')
plt.plot(wipro.rolling(window=30).mean(), color='#43AA8B',linestyle='--', label='30 Day M.A.')
plt.plot(wipro.rolling(window=90).mean(), color='#CBEF43',linestyle='-.', label='90 Day M.A.')
plt.title("Actual Price vs 30 Days MA vs 90 Days MA")
plt.xlabel("Date")
plt.ylabel("Price (in INR)")
plt.grid(True)
plt.legend()
plt.show()
```

```python
ete_return = eternal.pct_change(fill_method=None)
gfp_return = godfry.pct_change(fill_method=None)
ongc_return = ongc.pct_change(fill_method=None)
reliance_return = reliance.pct_change(fill_method=None)
wipro_return = wipro.pct_change(fill_method=None)

daily_returns = pd.DataFrame({
    "ETERNAL"  : ete_return,
    "GODFRY"   : gfp_return,
    "ONGC"     : ongc_return,
    "RELIANCE" : reliance_return,
    "WIPRO"    : wipro_return
})
daily_returns.head()
sns.histplot(data = daily_returns)
plt.title("Distribution of Daily Returns")
plt.xlabel("Daily Return")
plt.ylabel("Frequency")
plt.show()

print("Avg Return  :", np.mean(daily_returns["WIPRO"]))
print("Volatility  :", np.std(daily_returns["WIPRO"]))
print("Best Day    :", np.max(daily_returns["WIPRO"]))
print("Worst Day   :", np.min(daily_returns["WIPRO"]))
```

# Correlation Heatmap

```python
data = daily_returns.corr()
sns.heatmap(data, annot=True,cmap = "YlGnBu",fmt = ".2f")
plt.title("Correlation Between Stock Daily Returns")
plt.show()
```

# Risk vs Return Analysis

```python
risks = daily_returns.std()
returns = daily_returns.mean()
plt.figure(figsize=(10, 6))
plt.scatter(risks,returns,c="#DA4167")
for stock in daily_returns.columns:
    plt.annotate(stock, (risks[stock], returns[stock]),textcoords="offset points",xytext=(8, 5))
plt.grid(True)
plt.xlabel("Risk")
plt.ylabel("Return")
plt.title("Risk vs Return")
plt.show()
```
